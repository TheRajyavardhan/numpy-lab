# Random Numbers

```python
import numpy as np
# rand() give values in [0,1)
print(np.random.rand(5))
print(np.random.rand(2,3))
```

```python
# randn() give standard normal distribution
# mean = ~0 , Std = ~1
print(np.random.randn(5))
```

# Random Integers

```python
print(np.random.randint(1,10))
print(np.random.randint(1,10,size=5))
print(np.random.randint(1,10,size=(2,3)))
```

# Probability Distribution

```python
# Normal distribution
print(np.random.normal(loc=50,scale=5,size=5))

# Uniform distribution
print(np.random.uniform(0,100,size=5))

# Binomial distribution
print(np.random.binomial(n=10,p=0.5,size=5))
```

# Reproducibility(seed)

```python
np.random.seed(42)
print(np.random.rand(3))
```

# Shuffling Data

```python
# Shuffle array(in-place)
a = np.array([1,2,3,4,5])
np.random.shuffle(a)
print(a)
print(np.random.permutation(a))
```

# Random Sampling

```python
choice = np.random.choice([10,20,30,40],size=2)
print(choice)
# With probabilities:
choice = np.random.choice([0,1],size=10,p=[0.7,0.3])
print(choice)

```
