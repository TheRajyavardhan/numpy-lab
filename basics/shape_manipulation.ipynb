# Using reshape()

```python
import numpy as np

a = np.array([1, 2, 3, 4, 5, 6])
a = a.reshape(2,3)
print(a)
a = a.reshape(3,2)
print(a)
print(a.reshape(2, -1))
```

# Flattening Arrays


```python
A = np.array([[1,2,3],
              [4,5,6]])
# ravel():
print(A.ravel())
# flatten():
print(A.flatten())
```

# Transpose


```python
print(A.T)
print(A.shape)
print(A.T.shape)
```

# Axis Awareness

```python
A = np.array([[1,2,3],
              [4,5,6]])
print(A.sum())
print(A.sum(axis=0)) # vertical or column-wise
print(A.sum(axis=1)) # horizontal or row wise
```

# Stacking Arrays

```python
# Vertical stack or for adding row
print(np.vstack([A,[1,2,3]]))

# Horizontal stack or for adding column
print(np.hstack([[[0],[0]],A]))
```

# Splitting Arrays


```python
print(np.split(a,3)) # for 1-D array.
print(np.hsplit(A,3))
print(np.vsplit(A,2))
```
