```python
import numpy as np

A = np.array([[4, 1, 7],
              [2, 9, 3],
              [6, 5, 8]])
```

```python
# Q1
print(A[:,1:2])
```

```python
# Q2
B = A[0:2,1:3]
print(B)
```

```python
# Q3
B[0,0] = 100
print(A)
print(B)
# A changed because of B. As B and A share same memory address. Any change in B which is view of A, A will also change its value
```

```python
# Q4:
print(A.sum(axis=1))
```

```python
# Q5:
print(A.max(axis=0))
```

```python
# Q6:
sorted_indices = A.argsort(axis=1)
print(sorted_indices)
```

```python
# Q7:

print(np.take_along_axis(A,sorted_indices,axis=1))
```

```python
# Q8:
nums = np.array([10,20,30])
print(A+nums)
```

```python
sorted_col = A.argsort(axis=0)
print(np.take_along_axis(A,sorted_col,axis=0))
print(A)
```
