# Sorting Basics

```python
import numpy as np

a = np.array([40, 10, 30, 20])
print(np.sort(a)) # safe copy
print(a)
a.sort() # return modified original
print(a)
```

# Sorting 2D Arrays

```python
A = np.array([[30,10,20],
              [60,40,50]])
print(np.sort(A,axis=1))
print(np.sort(A,axis=0))
```

# Sorting by Index

```python
a = np.array([40,10,30,20])
idx = np.argsort(a)
print(idx)
print(a[idx])
```

# Sorting Real Pattern

```python
scores = np.array([88,92,79,95])
names = np.array(["A","B","C","D"])

order = np.argsort(scores)
print(names[order])
print(scores[order])
```

# Searching Basics

```python
x = np.array([10,20,30,40,50])
idx = np.where(x>25)
print(x[idx])
```

# Location of Non-Zero Elements

```python
a = np.array([0,5,0,10])
non_zero_idx = np.nonzero(a)
print(a[non_zero_idx])
```

# Membership Testing

```python
a = np.array([10,20,30,40])
b = np.array([20,40])

print(np.isin(a, b))
```

# Unique Values

```python
x = np.array([1,2,2,3,3,3,4,4,5,5,5,6,7])
print(np.unique(x))
vals, counts = np.unique(x, return_counts=True)
print(vals)
print(counts)
```

# Set Opertion

```python
a = np.arange(1,5)
b = np.arange(3,7)
print(a,b)
print(np.union1d(a,b))
print(np.intersect1d(a,b))
print(np.setdiff1d(a,b))
print(np.setxor1d(a,b))
```
