# Dot product

```python
import numpy as np
# check shape of a and b first.
a = np.array([1, 2, 3])
b = np.array([4,5,6])
print(np.dot(a,b))
print(a@b)
```

# Matrix Multiplication

```python
A = np.array([[1,2,3],
              [4,5,6]])
B = np.array([[1,2],
              [3,4],
              [5,6]])
print("is A and B compatilble: ",A.shape, B.shape)
print(A@B)
```

# Determinant

```python
sqr_mat = np.array([[2,1],
                    [1,3]])
print(np.linalg.det(sqr_mat))
```

# Inverse

```python
print(np.linalg.inv(sqr_mat))
```

# Solving Linear Equation

```python
b = np.array([[5],[6]])
x = np.linalg.inv(sqr_mat)@b # don't use this as it is slower
print(x)
x = np.linalg.solve(sqr_mat,b)
print(x)
```
