```python
import numpy as np

a = np.array([10, 20, 30, 40])
```

# Core Aggregation Functions

```python
print(a.sum())
print(a.mean())
print(a.min())
print(a.max())
print(np.median(a))
print(np.std(a))
print(np.var(a))
```

# Aggregation on 2D Arrays

```python
A = np.array([[10, 20, 30],
              [40, 50, 60]])
print(A.sum())
print(A.sum(axis=0))
print(A.sum(axis=1))
print(A.mean(axis=0))
print(A.mean(axis=1))
print(np.median(A,axis=1))
print(np.median(A,axis=0))
```

# Arg Functions

```python
a = np.array([10,50,20,90])
print(a.argmax()) # show index of max value
print(a.argmin()) # show index of min value
print(A.argmax(axis=0))
print(A.argmin(axis=1))
```

# Cumulative Operations

```python
a = np.array([10,20,30,40])

print(a.cumprod())
print(a.cumsum())
print(A.cumsum(axis=0))
print(A.cumprod(axis=0))
print(A.cumsum())
```

# NaN-Safe Aggregation (NaN = "Not a Number")

```python
x = np.array([10,10,np.nan,40])

print(np.nanmean(x))
print(x.mean())
print(np.nansum(x))
print(np.nanprod(x))
```
